# __Text Representation__

Text representation means converting human-readable text into a numerical format that a machine learning model can understand and process.
Feature extraction, text vectorization and text representation are same

### Why we need?
We need text representation / feature extraction because machine learning models work with numbers, not raw human language.

### Why it's so difficult?
Text representation or feature extraction is difficult because human language is complex, while machine learning models work with numerical data. A single word can have different meanings depending on its context, word order can change the meaning of a sentence, and synonyms can express the same idea using different words. Negation can also completely change the meaning, such as "I like this movie" and "I don't like this movie." In addition, informal language, spelling variations, slang, abbreviations, emojis, and languages like Roman Urdu make the process even more challenging. Therefore, the main challenge of text representation is to convert human language into numerical features while preserving the important information and meaning of the original text.

### What is the core idea?
Convert human language (text) into numerical features that a machine learning model can understand and use to learn patterns.

# __Important Terms:__

### __1. Corpus__
A corpus is the complete collection of text or documents that we are working with.
### __2. Vocabulary__
A vocabulary is the collection of unique words found across the entire corpus.
### __3. Document__
A document is an individual piece of text within a corpus. It could be a sentence, paragraph, review, article, or any other text.
### __3. Word__
A word is an individual unit of text inside a document.

# __1:One-Hot Encoding__

One-Hot Encoding is a technique that converts each unique word in a vocabulary into a binary vector containing 0s and 1s.

Each word gets its own unique position, and only that word's position contains 1. All other positions contain 0.

| Advantages                        | Disadvantages                   |
| --------------------------------- | ------------------------------- |
| Simple                            | High dimensionality             |
| Easy to implement                 | Sparse vectors                  |
| No false numerical relationship   | Doesn't capture meaning         |
| Works well for small vocabularies | Doesn't capture word similarity |
| Easy to understand                | OOV words can be a problem      |


```python
import numpy as np
from sklearn.preprocessing import OneHotEncoder

words = np.array(["I", "love", "NLP"]).reshape(-1, 1)

encoder = OneHotEncoder()

encoded = encoder.fit_transform(words)

print(encoded.toarray())
```
