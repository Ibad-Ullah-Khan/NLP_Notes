# __NLP__:
NLP = Natural Language Processing

It is a field of AI that helps computers understand, process, and work with human language.

For example, humans can understand:

"I really hate this person."

An NLP system can analyze this sentence and determine:

- What language is it?
- What is the sentiment?
- Is it abusive?
- Is it offensive?
- What is the meaning?
- Which words are important?

So, simply:

NLP = Making computers able to work with human language.
