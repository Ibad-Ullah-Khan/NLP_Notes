# __9 Tokenization__
Tokenization is the process of breaking a text into smaller units called tokens. A token can be a word, sentence, character, or subword.
### Types of Tokenization

1. **Word Tokenization** — Splits text into individual words.

2. **Sentence Tokenization** — Splits text into individual sentences.

3. **Character Tokenization** — Splits text into individual characters.

4. **Subword Tokenization** — Splits words into smaller meaningful parts. Commonly used in modern NLP models.

# __2 Using regular expression:__

```python
# Word Tokenization

import re

sentance = "i am happy!"
tokes = re.findall('[\w]+',sentance)
tokes
```

```python
# Sentance Tokenization
import re

text = "Hello! My name is Ibad. I am learning NLP, and it's really interesting! Do you like NLP? I think it's amazing; don't you agree?"
sentences = re.split(r'[.!?;]+', text)

print(sentences)
```

# __10 Stemming:__
Stemming is an NLP technique used to reduce words to their root form by removing prefixes or suffixes.

The resulting word is called a stem. The stem is not always a real English word.

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()


def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])


sample = "walk walks walking walked"
stem_words(sample)
```

```python
# Stemming is fast but sometimes produces incorrect or incomplete words.
# Lemmatization is slow but it produce english word at the end
# If you want to show output to the user then use Lemmatization it you are not then stemming is best for you
```

# __11 Lemmatization__
Lemmatization is an NLP technique that reduces a word to its base or dictionary form, called a lemma.

Unlike stemming, lemmatization tries to return a real meaningful word.

```python
import nltk
from nltk.stem import WordNetLemmatizer

wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations = "?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word", "Lemma"))
for word in sentence_words:
    print("{0:20}{1:20}".format(word, wordnet_lemmatizer.lemmatize(word)))
```
