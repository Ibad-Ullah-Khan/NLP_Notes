# __Pipline__:
A series of steps performed one after another to complete a task

# __NLP Pipline__:
A set of steps followed to bulid an end to end NLP software

NLP sofware consitsts of the following steps:
- Data Acquisition:
collecting the data you need for your project.

- Text Preparation:
taking raw text and making it clean and ready for NLP processing.

- Feature engineering:
creating or selecting useful information (features) from text that helps a machine learning model make better predictions.

- Modeling:
using a machine learning or deep learning algorithm to learn patterns from your prepared text data and make predictions.It alos consitst of two steps(1: Model Building, 2: Model Evaluation)

- Deployment:
making your trained NLP model available for real-world use.It further consist of (1: Deployment, 2: Monitoring, 3: Model Update)

# __2: Test Preparation__:

```python
Text Preparation
 ├── Cleaning
 │    ├── HTML / Tag cleaning
 │    ├── Emoji
 │    └── Spelling checker
 │
 ├── Basic preprocessing
 │    ├── Basic
 │    │    └── Tokenization
 │    │         ├── Sentence
 │    │         └── Word
 │    │
 │    └── Optimal
 │         ├── Stop word removal
 │         ├── Stemming
 │         ├── Removing digits, punctuations
 │         ├── Lowercasing
 │         └── Language detection
 │
 └── Advance preprocessing
      ├── POS tagging
      ├── Parsing
      └── Coreference resolution
```

# __3: Feature Engineering__:

```python
Feature Engineering
 ├── 1. Text Representation (Vectorization)
 │    ├── Traditional / Frequency-Based
 │    │    ├── One-Hot Encoding
 │    │    ├── Bag of Words (BoW)
 │    │    ├── N-grams (Unigram, Bigram, Trigram)
 │    │    └── TF-IDF (Term Frequency-Inverse Document Frequency)
 │    │
 │    ├── Static Embeddings
 │    │    ├── Word2Vec (CBOW, Skip-Gram)
 │    │    ├── GloVe
 │    │    └── FastText
 │    │
 │    └── Contextual Embeddings
 │         ├── BERT
 │         ├── RoBERTa
 │         └── Sentence Transformers (SBERT)
 │
 ├── 2. Feature Extraction (MetaData & Linguistics)
 │    ├── Character Level
 │    │    ├── Character Count
 │    │    ├── Word Count
 │    │    └── Punctuation / Special Character Count
 │    │
 │    ├── Syntactic / Structural Features
 │    │    ├── Part-of-Speech (POS) Tag Counts (Nouns, Verbs, etc.)
 │    │    ├── Sentence Length / Word Ratio
 │    │    └── Uppercase / Capitalization Count
 │    │
 │    └── Domain-Specific / Sentiment Features
 │         ├── Sentiment Polarity / Subjectivity Scores
 │         ├── Readability Scores (Flesch-Kincaid, etc.)
 │         └── Domain Keyword Flags / Topic Vectors
 │
 ├── 3. Feature Transformation & Scaling
 │    ├── Normalization (Min-Max Scaling)
 │    ├── Standardization (Z-score Scaling)
 │    └── Log Transformation (for skewed distributions)
 │
 └── 4. Feature Selection & Dimensionality Reduction
      ├── Filter Methods
      │    ├── Chi-Square Test ($\chi^2$)
      │    ├── Information Gain / Mutual Information
      │    └── Variance Thresholding
      │
      ├── Wrapper / Embedded Methods
      │    ├── Recursive Feature Elimination (RFE)
      │    └── L1 Regularization (Lasso)
      │
      └── Dimensionality Reduction
           ├── Principal Component Analysis (PCA)
           ├── Truncated SVD (LSA - Latent Semantic Analysis)
           └── t-SNE / UMAP (Visualization)
```
